```sql
CREATE OR REPLACE TABLE alan_customers_bronze AS
SELECT 
  CAST(customer_id AS STRING) AS customer_id,
  CAST(name AS STRING) AS name,
  CAST(email AS STRING) AS email,
  to_date(dob, 'yyyy-MM-dd') AS dob,
  to_date(signup_date, 'yyyy-MM-dd') AS signup_date,
  CAST(location_id AS STRING) AS location_id,
  current_timestamp() AS ingestion_time,
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_customers.csv' AS source_file_name
FROM read_files(
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_customers.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
SELECT * FROM alan_customers_bronze;
```

```sql
CREATE OR REPLACE TABLE alan_restaurants_bronze AS
SELECT
  CAST(restaurant_id AS STRING) AS restaurant_id,
  CAST(name AS STRING) AS name,
  CAST(cuisines AS STRING) AS cuisines,
  CAST(location_id AS STRING) AS location_id,
  CAST(rating AS DOUBLE) AS rating,
  CAST(delivery_fee AS DOUBLE) AS delivery_fee,
  current_timestamp() AS ingestion_time,
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_restaurants.csv' AS source_file_name
FROM read_files(
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_restaurants.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
SELECT * FROM alan_restaurants_bronze;
```

```sql
CREATE OR REPLACE TABLE alan_locations_bronze AS
SELECT
  CAST(location_id AS STRING) AS location_id,
  CAST(area AS STRING) AS area,
  CAST(city AS STRING) AS city,
  CAST(state AS STRING) AS state,
  CAST(latitude AS DOUBLE) AS latitude,
  CAST(longitude AS DOUBLE) AS longitude,
  current_timestamp() AS ingestion_time,
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_locations.csv' AS source_file_name
FROM read_files(
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_locations.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
SELECT * FROM alan_locations_bronze;
```

```sql
CREATE OR REPLACE TABLE alan_delivery_agents_bronze AS
SELECT
  CAST(agent_id AS STRING) AS agent_id,
  CAST(name AS STRING) AS name,
  CAST(phone AS STRING) AS phone,
  CAST(rating AS DOUBLE) AS rating,
  current_timestamp() AS ingestion_time,
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_delivery_agents.csv' AS source_file_name
FROM read_files(
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_delivery_agents.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
SELECT * FROM alan_delivery_agents_bronze;
```

```sql
CREATE OR REPLACE TABLE alan_menu_items_bronze AS
SELECT
  CAST(item_id AS STRING) AS item_id,
  CAST(restaurant_id AS STRING) AS restaurant_id,
  CAST(item_name AS STRING) AS item_name,
  CAST(price AS DOUBLE) AS price,
  CAST(category AS STRING) AS category,
  current_timestamp() AS ingestion_time,
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_menu_items.csv' AS source_file_name
FROM read_files(
  '/Volumes/alankrit_db_workspace_7405606622859005/alan_dinedash_db/alan_dinedash_raw/dim_menu_items.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
SELECT * FROM alan_menu_items_bronze;
```

```sql
CREATE OR REPLACE TABLE alan_customers_filtered AS
SELECT * FROM alan_customers_bronze
WHERE customer_id IS NOT NULL 
  AND location_id IS NOT NULL 
  AND signup_date IS NOT NULL 
  AND dob IS NOT NULL 
  AND email IS NOT NULL 
  AND name IS NOT NULL;
```

```sql
SELECT * FROM alan_customers_filtered;
```

```sql
SELECT 'Bronze' AS layer, COUNT(*) AS count FROM alan_customers_bronze
UNION ALL
SELECT 'Filtered' AS layer, COUNT(*) AS count FROM alan_customers_filtered;
```
